# LES CATHERINES

Les Catherines are a trio blending the warm sounds of acoustic instruments with a rich mix of languages (English, French, Spanish, and more). Their eclectic style, inspired by world music and soulful interpretations, creates a captivating, multicultural listening experience. Each performance is an invitation to travel across languages, emotions, and musical traditions.

![Les Catherines live](images/les-catherines-live.jpg)


## Live Performances

As a trio, we perform regularly across Boulder County beyond, bringing our music to intimate venues, restaurants and outdoor spaces.

[See upcoming events →](events/#performances)

## Music Community

In addition to our performances, we organize and run the CBArt Studios Open Mic in Superior. CBArt Studios Open Mic is a warm and welcoming space where local musicians and audiences come together in a shared musical experience.
 
[More about the CBArt Studios Open Mic →](events/#openmic)
