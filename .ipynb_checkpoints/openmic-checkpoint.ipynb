# CBart Open Mic

In addition to performing as a band, Les Catherines organize and host the CBArt Studios Open Mic. This is a community event that gives local musicians a welcoming and supportive place to share their music.

## An open mic hosted by Les Catherines

::::{grid} 2
:gap: 2
:class: responsive-grid
:::{grid-item}

(openmic)=

Thanks to the generosity of **CBArt Studios**, which provides its beautiful space for live music, the Open Mic takes place on the second Friday of every other month. CBArt Studios also offers complimentary snacks and drinks for performers and guests.

The evening is all about sharing music, discovering new voices, and creating connections within the local community.
