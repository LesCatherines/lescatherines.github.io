## CBArt Studios Open Mic

Thanks to the generosity of CBArt Studios, the Open Mic takes place in their beautiful space on the second Friday of every other month.

Musicians of different styles and experience levels are welcome. The evening is about sharing music, discovering new voices, and connecting with the local music community.

CBArt Studios also provides complimentary snacks and drinks for performers and guests.
