# EVENTS

Les Catherines regularly perform as a trio across Boulder County and the Denver area. Alongside these performances, we also foster a local music community through the CBArt Studios Open Mic in Superior.

## Les Catherines Performances

We perform regularly in Boulder County, including CBArt Studios in Superior and Tandoori Grill in Boulder. We have also been invited to perform at venues such as the Alliance Française de Denver and the Cherry Creek Shopping Center. Our performances feature a multilingual acoustic repertoire, blending voices, styles, and interpretations into a warm and engaging musical experience.

## Music Community — CBArt Studios Open Mic

In addition to performing, we organize and run the CBArt Studios Open Mic in Superior, a recurring gathering where musicians and listeners come together in a relaxed and supportive setting.

Thanks to the generosity of CBArt Studios, which provides their beautiful space for live music, the open mic takes place on the first Friday of every other month. It is an evening dedicated to sharing music, discovering new voices, and creating connections within the local community.
