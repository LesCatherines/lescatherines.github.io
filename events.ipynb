# EVENTS

Les Catherines regularly perform as a trio in intimate venues, restaurants and outdoor spaces, across Boulder County and the Denver area. Alongside these performances, we also foster a local music community through the CBArt Studios Open Mic in Superior.

## Les Catherines Performances

Our performances feature a multilingual acoustic repertoire, blending voices, styles, and interpretations into a warm and engaging musical experience. Recent venues include: CBArt Studios in Superior, Tandoori Grill in Boulder, the Alliance Française in Denver and the Cherry Creek Shopping Center.

## Music Community — CBArt Studios Open Mic

In addition to performing, we organize and run the CBArt Studios Open Mic in Superior, a recurring gathering where musicians and listeners come together in a relaxed and supportive setting.

Thanks to the generosity of CBArt Studios, which provides their beautiful space for live music, the open mic takes place on the first Friday of every other month. It is an evening dedicated to sharing music, discovering new voices, and creating connections within the local community.
