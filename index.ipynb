# LES CATHERINES

Les Catherines is an acoustic trio with an **eclectic repertoire inspired by music from around the world**.

Singing in English, French, Spanish, and more, we create our own acoustic arrangements of songs we love, from familiar favorites to unexpected discoveries. 

Each performance is a musical journey across languages, emotions, and musical traditions.

![Les Catherines live](images/les-catherines-live.jpg)


## Live Performances

We perform regularly across Boulder County and the Denver area, in intimate venues, restaurants, festivals, and community spaces.
