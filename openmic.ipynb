## CBArt Studios Open Mic

The CBArt Studios Open Mic takes place on the **second Friday of every other month**. Musicians of different styles and experience levels are welcome. 

The evening is about sharing music, discovering new voices, and connecting with the local music community.

We are grateful to **[CBArt Studios](https://cbartstudios.com/)** for opening its beautiful space to live music and providing complimentary snacks and drinks for performers and their guests.
