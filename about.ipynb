
# ABOUT US

---

![Les Catherines live](images/les-catherines-root.jpg)



Les Catherines is an acoustic trio born from a simple question: **“We really like that song. What could we do with it?”**

That question is at the heart of what we do. We start with songs we love and explore them until we find an arrangement that feels right for our voices and instruments. Sometimes it comes together quickly; sometimes it takes a while. And even when we think we are finished, we may change our minds again.

Our repertoire moves between well-known songs such as *Feeling Good*, *Wish You Were Here*, and *In My Life*, along with songs that may be less familiar to our audiences. We move across languages, cultures, and musical styles, letting each song find its own mood and character.

For us, the fun is in finding a version of the song that feels like ours. Sometimes it stays very close to where it started; sometimes it goes somewhere else entirely.


## MEET LES CATHERINES

**Patrizia — Vocals · Québec** </br>

Patrizia's powerful and expressive voice is at the heart of Les Catherines' sound. Her gift for languages lets us explore songs in many different languages.

**Vincent — Guitar · Belgium**</br>

Vincent's guitar shapes much of the trio's acoustic sound. He brings creative ideas to the arrangements, and to how we make each song our own.

**Cecile — Bass/Vocals. Belgium**</br>

Cecile wears many hats in the trio, moving between bass, vocals, and ukulele to add variety to the music, while also taking care of the band's sound.
