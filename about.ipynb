
# ABOUT US

---

![Les Catherines live](images/les-catherines-root.jpg)



Les Catherines is an acoustic trio born from a simple question: “We really like that song. What could we do with it?”

We don’t just play covers; we listen, experiment, explore arrangements until something feels right to us for our voices and instruments. Sometimes an arrangement comes together quickly; other times it takes a while. But when it finally lands, it tends to stay... until we change our minds again.

Our repertoire includes well-known songs (such as *Feeling Good*, *Wish You Were Here*, and *In My Life*), alongside much lesser-known pieces that we enjoy bringing into the light. We let each song find the language, color, and atmosphere that suit it best.

The result is a varied repertoire in which each song carries its own atmosphere. We are after a shared moment of emotion, connection, and surprise (a listener recognizes a song and hears it differently at the same time).
